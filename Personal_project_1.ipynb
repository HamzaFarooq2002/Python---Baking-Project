```python
#Python Banking project

def show_balance(balance):
    print(f"Your Balance is Rs.{balance:.2f}")

def deposit():
    amount = float(input("Please enter the amount you would like to deposit: "))
    if amount < 0:
        print("This is an invalid entry")
        return 0  # Return 0 instead of None
    else:
        return amount

def withdraw(balance):
 cashout=float(input("Enter the amount you want to withdraw"))
 if cashout>balance:
  print("Insufficient funds")
  return 0
 elif cashout<0:
  print("Invalid entry")
  return 0
 else:
  return cashout

def main():
 balance = 10
 is_running = True

 while is_running:
    print("Welcome to Python Banking Program")
    print("Menu:")
    print("1. Show Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")
    choice = input("Enter your choice: ")

    if choice == "1":
        show_balance(balance)
    elif choice == "2":
        balance += deposit()
    elif choice == "3":
        balance -= withdraw(balance)
    elif choice == "4":
        is_running = False
    else:
        print("I am sorry, this input is not valid. Please try again.")

print("\nThank you, have a nice day!")
if __name__ == '__main__':
 main()
```
